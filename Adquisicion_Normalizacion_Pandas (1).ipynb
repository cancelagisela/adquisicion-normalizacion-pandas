```python
productos = pd.DataFrame({
"id_producto": [101, 102, 103, 104],
"nombre_producto": [
"Teclado",
"Ratón",
"Monitor",
"Auriculares"
],
"precio_unitario": [
25.50,
15.00,
180.00,
40.00
]
})
```

```python
productos.to_excel(
"productos.xlsx",
index=False
)
```

```python
transacciones = pd.read_csv("transacciones.csv")

productos = pd.read_excel(
"productos.xlsx",
engine="openpyxl"
)

print(transacciones.head())
print()
print(productos.head())
```

```python
print(transacciones.dtypes)
print()
print(productos.dtypes)
```

```python
transacciones["fecha"] = pd.to_datetime(
transacciones["fecha"]
)

print(transacciones.dtypes)
```

```python
print(transacciones.isnull().sum())
```

```python
transacciones["cantidad"] = (
transacciones["cantidad"].fillna(0)
)
```

```python
print(transacciones.isnull().sum())
```

```python
ventas = pd.merge(
transacciones,
productos,
on="id_producto",
how="left"
)
```

```python
ventas["total_venta"] = (
ventas["cantidad"] *
ventas["precio_unitario"]
)
```

```python
print(ventas.isnull().sum())
```

```python
ventas.to_parquet(
"ventas_normalizadas.parquet",
index=False
)
```
